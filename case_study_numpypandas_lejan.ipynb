 Import pandas, numpy

```python
import numpy as np
import pandas as pd
```

Load dataset using pd.read_csv()  

```python
filepath='https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'
df=pd.read_csv(filepath)
df
```

 Check and remove duplicates

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
df
```

 Rename columns ("Title" > "Movie_Title")

```python
df=df.rename(columns={'Title':'Movie_Title'})
df.columns
```

Get unique genres

```python
print(df["Genre"].unique())
```

 Extract number of genres per movie


```python
df["Genre_Count"] = df["Genre"].str.split("|").str.len()
print(df[["Movie_Title", "Genre", "Genre_Count"]].head())
```

Handle missing values in 'Revenue (Millions)' and 'Metascore

```python
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(df["Revenue (Millions)"].median())
df["Metascore"] = df["Metascore"].fillna(df["Metascore"].median())
print(df[["Revenue (Millions)", "Metascore"]].isnull().sum())
```

Top 5 movies by Rating

```python
top_5 = df.sort_values("Rating", ascending=False).head(5)
print(top_5["Movie_Title"])
```

Movies with highest/lowest revenue

```python
highest_revenue = df.loc[df["Revenue (Millions)"].idxmax()]
print(highest_revenue["Movie_Title"])

lowest_revenue = df.loc[df["Revenue (Millions)"].idxmin()]
print(lowest_revenue['Movie_Title'])
```

 Avg. rating by Director

```python
avg_rating_director = df.groupby("Director")["Rating"].mean()

print(avg_rating_director)
```

 Movies per year

```python
movies_per_year = df.groupby("Year").size()
print(movies_per_year)
```

Movies with rating > 8 & revenue > 100M

```python
high_rated = df[(df["Rating"] > 8) &  (df["Revenue (Millions)"] > 100)]
print(high_rated['Movie_Title'])
```

 Group by Director: avg Rating, Revenue

```python
director_stats = df.groupby("Director")[["Rating", "Revenue (Millions)"]].mean()
print(director_stats)
```

Group by Year: count, avg Rating

```python
year_stats = df.groupby("Year").agg(Movie_Count=("Movie_Title", "count"),Average_Rating=("Rating", "mean"))
print(year_stats)
```

Create a new column Profitability:
a) Revenue divided by Duration
b) Find top 10 most profitable movies

```python
df["Profitability"] = (df["Revenue (Millions)"] / df["Runtime (Minutes)"])
print(df[["Movie_Title", "Revenue (Millions)", "Runtime (Minutes)", "Profitability"]].head())

top_10_profit = df.sort_values("Profitability",ascending=False).head(10)
print(top_10_profit["Movie_Title"])
```

 Create a column that categorizes movies by rating:
a) Rating < 6: Poor
b) 6 ≤ Rating < 7.5: Average
c) Rating ≥ 7.5: Excellent

```python
conditions = [df["Rating"] < 6, df["Rating"] < 7.5]
choices = ["Poor","Average"]
df["Rating_Category"] = np.select(conditions,choices,default="Excellent")
print(df[["Movie_Title", "Rating", "Rating_Category"]].head(10))
```

Create a custom function to score movies based on:
i)
Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score
Apply a function to dataset to assign the score


```python
def movie_score(row):
    return row["Rating"] * 10 + row["Revenue (Millions)"]

df["Movie_Score"] = df.apply(movie_score, axis=1)
print(df[["Movie_Title", "Rating", "Revenue (Millions)", "Movie_Score"]].head())
```
